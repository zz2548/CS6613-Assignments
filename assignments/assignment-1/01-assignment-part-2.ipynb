# Solving the estimation problem

We have a 2-component Mixture of Gaussians (MoG) with 1-dimensional data.
The likelihood for a point $x_i$ is:

$p(x_i  \mid \pi, \mu, \sigma^2) = \pi_1 \, \mathcal{N}(x_i \mid \mu_1, \sigma_1^2) + \pi_2 \, \mathcal{N}(x_i \mid \mu_2, \sigma_2^2).$

The log-likelihood for $m$ data points is:

$\mathcal{L} = \sum_{i=1}^m \log \left(
\sum_{k=1}^2 \pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)
\right).$


Define the responsibility of component $k$ for data point $i$:

$\gamma_{ik} =
\frac{ \pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) }
     { \sum_{j=1}^2 \pi_j \, \mathcal{N}(x_i \mid \mu_j, \sigma_j^2) }.
     $

The question asks us to prove that the gradients given are needed for solving the estimation problem.

- Mean gradient:
$\frac{\partial \mathcal{L}}{\partial \mu_k}
= \sum_{i=1}^m \gamma_{ik} \, \frac{(x_i - \mu_k)}{\sigma_k^2}.$

- Variance gradient:
$\frac{\partial \mathcal{L}}{\partial \sigma_k^2}
= \frac{1}{2} \sum_{i=1}^m \gamma_{ik}
  \left[ \frac{(x_i - \mu_k)^2}{\sigma_k^4} - \frac{1}{\sigma_k^2} \right].$

- Mixture weights gradient(with constraint $\sum_k \pi_k = 1$):
$\frac{\partial \mathcal{L}}{\partial \pi_k}
= \sum_{i=1}^m \frac{\gamma_{ik}}{\pi_k}.$

## Part 1: Mean Gradient

Starting with the mean gradient:
Solve for $ \frac {\partial \mathcal{L}}{\partial \mu_k}$

Using the log-likelihood formula:
take partial derivative of both sides and divide by $ \partial \mu_k$

$$\mathcal{L} = \sum_{i=1}^m \log \left(
\sum_{k=1}^2 \pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)
\right).$$

$$\frac{\partial \mathcal{L}}{\partial \mu_k} = \sum_{i=1}^m \frac{\partial}{\partial \mu_k} \log \left( \sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2) \right)$$

Apply rule:
$ \frac{d}{dx} log (f(x)) = \frac{1}{f(x)} \cdot \frac{df}{dx}$



$$\frac{\partial \mathcal{L}}{\partial \mu_k} = \sum_{i=1}^m \frac{1}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)} \cdot \frac{\partial}{\partial \mu_k} \left( \sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2) \right)$$

Looking at the last part of our partial derivative:
The term that contains $ \mu_k $ is non-zero.
Therefore:

$$\frac{\partial}{\partial \mu_k}\left( \sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)\right) = \pi_k \frac{\partial}{\partial \mu_k} \mathcal{N} (x_i \mid \mu_k, \sigma_k^2)$$



We plug it back in:
$$\frac{\partial \mathcal{L}}{\partial \mu_k} = \sum_{i=1}^m \frac{\pi_k \frac{\partial}{\partial \mu_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

Need to find: $\frac{\partial}{\partial \mu_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)$ on the top

$$\frac{\partial}{\partial \mu_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)$$

$$=\mathcal{N}(x_i \mid \mu_k, \sigma_k^2) \cdot \frac{\partial}{\partial \mu_k} \left(-\frac{(x_i-\mu_k)^2}{2\sigma_k^2}\right) $$

$$=  \frac{1}{\sqrt{2 \pi \sigma_k^2 }}exp(-\frac {(x_i-\mu_k)^2}{2 \sigma^2_k}) \cdot \frac{x_i - \mu_k}{\sigma_k^2} $$

$$=  \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) \cdot \frac{x_i - \mu_k}{\sigma_k^2} $$

We plug it back in:
$$\frac{\partial \mathcal{L}}{\partial \mu_k} = \sum_{i=1}^m \frac{ \pi_k \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) }{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)} \cdot \frac{x_i - \mu_k}{\sigma_k^2}$$

The given responsibility function is: $$\gamma_{ik} =
\frac{ \pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) }
     { \sum_{j=1}^2 \pi_j \, \mathcal{N}(x_i \mid \mu_j, \sigma_j^2) }.$$

Therefore we have arrived at the mean gradient: $$\boxed{\frac{\partial \mathcal{L}}{\partial \mu_k}
= \sum_{i=1}^m \gamma_{ik} \, \frac{(x_i - \mu_k)}{\sigma_k^2}}$$

## Part 2: Variance Gradient

Need to find variance gradient $\frac{\partial \mathcal{L}}{\partial \sigma_k^2}$

Using this from part 1: $$\frac{\partial \mathcal{L}}{\partial \mu_k} = \sum_{i=1}^m \frac{\pi_k \frac{\partial}{\partial \mu_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

Becomes:$$\frac{\partial \mathcal{L}}{\partial\sigma^2_k} = \sum_{i=1}^m \frac{\pi_k \frac{\partial}{\partial \sigma^2_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

Given the Gaussian density: $$\mathcal{N}(x_i \mid \mu_k, \sigma_k^2) = \frac{1}{\sqrt{2 \pi \sigma_k^2 }}exp(-\frac {(x_i-\mu_k)^2}{2 \sigma^2_k})$$


Take derivative with respect to $ \sigma^2_k $
$$ \frac{\partial}{\partial\sigma^2_k} \left (\frac{1}{\sqrt{2 \pi \sigma_k^2 }} \right )$$

$$ = - \frac {1}{2} (2 \pi \sigma^2_k)^{-3/2} \cdot 2 \pi $$

$$= \frac{1}{2 \sigma^2_k} (2\pi\sigma^2_k)^{-1/2}$$

$$ \frac{\partial}{\partial\sigma^2_k} \left ( exp(-\frac {(x_i-\mu_k)^2}{2 \sigma^2_k}) \right)$$

$$ = exp(-\frac {(x_i-\mu_k)^2}{2 \sigma^2_k}) \cdot \frac {(x_i-\mu_k)^2}{2 \sigma^4_k}$$

Combine into:

$$\frac{\partial}{\partial\sigma^2_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) = \frac{1}{2 \sigma^2_k} (2\pi\sigma^2_k)^{-1/2} \cdot exp(-\frac {(x_i-\mu_k)^2}{2 \sigma^2_k}) \cdot \frac {(x_i-\mu_k)^2}{2 \sigma^4_k}
$$

We know that $$\mathcal{N}(x_i \mid \mu_k, \sigma_k^2) = \frac{1}{\sqrt{2 \pi \sigma_k^2 }}exp(-\frac {(x_i-\mu_k)^2}{2 \sigma^2_k})$$


Simplify into: 
$$\frac{\partial}{\partial\sigma^2_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) = \mathcal{N}(x_i \mid \mu_k, \sigma_k^2) \left [ - \frac{1}{2 \sigma^2_k} + \frac {(x_i-\mu_k)^2}{2 \sigma^4_k} \right]$$

Substitute the above equation into: $$\frac{\partial \mathcal{L}}{\partial\sigma^2_k} = \sum_{i=1}^m \frac{\pi_k \frac{\partial}{\partial \sigma^2_k} \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

$$\frac{\partial \mathcal{L}}{\partial\sigma^2_k} 
= \sum_{i=1}^m 
\frac{\pi_k \frac{\partial}{\partial \sigma^2_k} 
\mathcal{N}(x_i \mid \mu_k, \sigma_k^2) \left [ - \frac{1}{2 \sigma^2_k} + \frac {(x_i-\mu_k)^2}{2 \sigma^4_k} \right]}
{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

$$\frac{\partial \mathcal{L}}{\partial\sigma^2_k} 
= \sum_{i=1}^m 
\frac{\pi_k \frac{\partial}{\partial \sigma^2_k} 
\mathcal{N}(x_i \mid \mu_k, \sigma_k^2) }
{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}
\cdot
\left [ - \frac{1}{2 \sigma^2_k} + \frac {(x_i-\mu_k)^2}{2 \sigma^4_k} \right]$$

We notice the left side is the responsibility function $\gamma_{ik}$:

$$\frac{\partial \mathcal{L}}{\partial\sigma^2_k} 
= \gamma_{ik}
\cdot
\left [ - \frac{1}{2 \sigma^2_k} + \frac {(x_i-\mu_k)^2}{2 \sigma^4_k} \right]$$

Take out the constant $ \frac {1}{2}$, we arrive at the variance gradient
$$
\boxed{
\frac{\partial \mathcal{L}}{\partial \sigma_k^2}
= \frac{1}{2} \sum_{i=1}^m \gamma_{ik}
  \left[ \frac{(x_i - \mu_k)^2}{\sigma_k^4} - \frac{1}{\sigma_k^2} \right]
}
  $$

## Part 3: Mixture weight gradient(with constraint $\sum_k \pi_k = 1$)

Use the  log-likelihood:


$\mathcal{L} = \sum_{i=1}^m \log \left(
\sum_{k=1}^2 \pi_k \, \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)
\right).$

Take derivative with respect to $pi_k$

We want to find $\frac{\partial \mathcal{L}}{\partial \pi_k}$ (unconstrained).

$$\frac{\partial \mathcal{L}}{\partial \pi_k} = \sum_{i=1}^m \frac{\frac{\partial}{\partial \pi_k} \left( \sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2) \right)}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

Only the $k$-th term in the sum depends on $\pi_k$:
$$\frac{\partial \mathcal{L}}{\partial \pi_k} = \sum_{i=1}^m \frac{\mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}{\sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$

Multiplying both top and bottom by $\pi_k$:
$$\frac{\partial \mathcal{L}}{\partial \pi_k} = \sum_{i=1}^m \frac{\pi_k \mathcal{N}(x_i \mid \mu_k, \sigma_k^2)}{\pi_k \sum_{j=1}^2 \pi_j \mathcal{N}(x_i \mid \mu_j, \sigma_j^2)}$$



$$\boxed{\frac{\partial \mathcal{L}}{\partial \pi_k} = \sum_{i=1}^m \frac{\gamma_{ik}}{\pi_k}}$$

We have shown that the gradients we need to solve the estimation problem are the boxed equations above
